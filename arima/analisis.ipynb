```python
import matplotlib.pyplot as plt
```

```python
series = []
with open("data/procesos_arima.csv", "r") as archivo_series:
    for linea in archivo_series:
        series.append([float(x) for x in linea.strip().split(",") if x != ""])
```

```python
def grafico_serie(serie, i):
    plt.figure(figsize=(10, 6))
    plt.plot(serie)
    plt.xlabel("Tiempo")
    plt.ylabel("Valor")
    plt.title(f"Serie {i + 1}")
    plt.show()
```

```python
grafico_serie(series[0], 0)
```

```python
grafico_serie(series[1], 1)
```

```python
grafico_serie(series[2], 2)
```

```python
grafico_serie(series[3], 3)
```

```python
grafico_serie(series[4], 4)
```

```python
grafico_serie(series[5], 5)
```

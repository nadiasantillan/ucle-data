```python
from common import Tendencia, Autocorrelacion, Serie, Estacionaridad
from common import Cubo
import pandas as pd
from scipy.signal import find_peaks


```

# Análisis Exploratorio

```python
vegetacion_cubo = Cubo("data/vegetacion.csv", "data/tiempo.csv")
temperatura_cubo = Cubo("data/temperaturas.csv", "data/tiempo.csv")
```

## Graficos

```python
for i in range(3):
    for j in range(3):
        serie = Serie(
            vegetacion_cubo.tiempo, 
            vegetacion_cubo.datos[i, j, :], 
            temperatura_cubo.datos[i, j, :], 
            f"Temperatura y Vegetación [{i}][{j}]",
            "Vegetación",
            "Temperatura"
            )
        serie.grafico()
```

## Análisis de tendencia

```python
for i in range(3):
    for j in range(3):
        analisis_tendencia = Tendencia(vegetacion_cubo.tiempo, vegetacion_cubo.datos[i, j,:], f"Vegetacion [{i}][{j}]")
        analisis_tendencia.grafico()
```

## Análisis de Estacionalidad Vegetación y Temperatura

```python
for i in range(3):
    for j in range(3):
        autocorrelacion = Autocorrelacion(vegetacion_cubo.tiempo, vegetacion_cubo.datos[i, j,:], f"Vegetacion [{i}][{j}]")
        autocorrelacion.grafico()
        picos, _ = find_peaks(autocorrelacion.acf())
        print(pd.DataFrame({"Indices": picos, "ACF": autocorrelacion.acf()[picos]}))

        autocorrelacion_temp = Autocorrelacion(temperatura_cubo.tiempo, temperatura_cubo.datos[i, j,:], f"Temperatura [{i}][{j}]")
        autocorrelacion_temp.grafico()
        picos_temp, _ = find_peaks(autocorrelacion_temp.acf())
        print(pd.DataFrame({"Indices": picos_temp, "ACF": autocorrelacion_temp.acf()[picos_temp]}))
        
```

```python
for i in range(3):
    for j in range(3):
        estacionaridad_vegetacion = Estacionaridad(vegetacion_cubo.datos[i, j, :], f"Estacionaridad Vegetacion [{i}][{j}]")
        estacionaridad_temperatura = Estacionaridad(temperatura_cubo.datos[i, j, :], f"Estacionaridad Vegetacion [{i}][{j}]")
        
        print(f"ADF vegetación corto plazo[{i}][{i}]\n",estacionaridad_vegetacion.adf_corto_plazo())
        print(f"ADF vegetación diferencia estacional[{i}][{i}]\n",estacionaridad_vegetacion.adf_largo_plazo())
        print(f"Confirmacion raíz estacional vegetacion[{i}][{i}]", estacionaridad_vegetacion.confirmacion_raiz_estacional() )
        print(f"ADF temperatura corto plazo[{i}][{i}]\n",estacionaridad_temperatura.adf_corto_plazo())
        print(f"ADF temperatura diferencia estacional[{i}][{i}]\n",estacionaridad_temperatura.adf_largo_plazo())        
        print(f"Confirmacion raíz estacional temperatura[{i}][{i}]", estacionaridad_temperatura.confirmacion_raiz_estacional() )
```

```python
from common import Cubo
import matplotlib.pyplot as plt
```

```python
vegetacion_cubo = Cubo("data/vegetacion.csv", "data/tiempo.csv")
temperatura_cubo = Cubo("data/temperaturas.csv", "data/tiempo.csv")
```

```python
def comparar(serie1, serie2):
    iguales = True
    for i in range(len(serie1)):
        if serie1[i] != serie2[i]:
            iguales = False
            break
    return iguales
```

Las series de temperaturas 00, 01, 02 y 12 son iguales. Lo mismo sucede con las series 10, 11, 20, 21 y 22

```python
for i in range(3):
    for j in range(3):
        for k in range(i, 3):
            for l in range(j, 3):
                if (i != k or j != l) and (k > i or l > j):
                    iguales = comparar(temperatura_cubo.datos[i, j, :], temperatura_cubo.datos[k, l, :])
                    print(f"Comparación Temperatura {i}{j} y Temperatura {k}{l}: {'Iguales' if iguales else 'Diferentes'}")
        
```

```python
plt.figure(figsize=(15, 6))
for i, j in [(0, 0), (0, 1), (0, 2), (1, 2)]:
    plt.plot(temperatura_cubo.tiempo, temperatura_cubo.datos[i, j, :], label=f"Temperatura {i}{j}")
    plt.xlabel("Tiempo")
    plt.ylabel("Temperatura")
    plt.title("Temperaturas")
    plt.legend()
plt.show()
```
